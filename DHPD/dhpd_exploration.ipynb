# Under Implementation Projects:

```python

from pathlib import Path

def load_excel_files_as_dfs(folder):
    """Load Excel files into notebook variables named df1, df2, and so on."""
    folder = Path(folder)
    loaded = {}

    files = sorted(
        file for file in folder.iterdir()
        if file.is_file() and file.suffix.lower() in {".xlsx", ".xls"}
    )

    for i, file in enumerate(files, start=1):
        name = f"df{i}"
        globals()[name] = pd.read_excel(file)
        loaded[name] = globals()[name]

    return loaded


loaded_dfs = load_excel_files_as_dfs("under_implementation")
```

```python
loaded_dfs.keys()
```

```python
siang = loaded_dfs['df1']
subansiri = loaded_dfs['df2']
lohit = loaded_dfs['df3']
dibang = loaded_dfs['df4']
kameng =loaded_dfs['df5']
tawang = loaded_dfs['df6']
dikrong = loaded_dfs['df7']
tirap = loaded_dfs['df8']
```

```python
projects_under_implementation = pd.concat([siang, subansiri, lohit, dibang, kameng, tawang, dikrong, tirap], ignore_index=True)
projects_under_implementation.info()
```

```python
def remove_commas_and_convert_to_float(df, column):
    df[column] = pd.to_numeric(
        df[column].astype("string").str.replace(",", "", regex=False)
    )
    return df

remove_commas_and_convert_to_float(projects_under_implementation, 'capacity_(mw)')
projects_under_implementation.drop(columns=['no. of projects'], inplace=True)
projects_under_implementation.info()
```

```python
under_implementation_capacity = projects_under_implementation['capacity_(mw)'].sum()
under_implementation_count = projects_under_implementation['capacity_(mw)'].count()
print(f'Total capacity of projects under implementation: {under_implementation_capacity} MW')
print(f'Number of projects under implementation: {under_implementation_count}')
```

# Under Construction Projects:

```python
projects_under_construction = pd.read_excel('under_construction/dhpdap-online-2026-11-14-6(1).xlsx')
projects_under_construction.info()
```

```python
projects_under_construction.drop(columns=['no. of projects'], inplace=True)
projects_under_construction.head()
```

```python
under_construct_capacity = projects_under_construction['capacity_(mw)'].sum()
under_construct_count = projects_under_construction['capacity_(mw)'].count()
print(f'Total capacity of projects under construction: {under_construct_capacity} MW')
print(f'Number of projects under construction: {under_construct_count}')
```

```python
# Overall total capacity
print(f'Total Overall Capacity: {round(under_implementation_capacity + completed_capacity + under_construct_capacity, 2)} MW, Total Number of Projects: {under_implementation_count + completed_projects_count + under_construct_count}')
```
